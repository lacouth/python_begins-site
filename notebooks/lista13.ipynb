# Lista 13 · Decompor e testar

A última lista do semestre. O formato é o mesmo da Lista 12 — função com
testes prontos no fim do arquivo —, mas alguns exercícios agora pedem
**duas** funções no mesmo arquivo, uma chamando a outra: é a decomposição do
[Capítulo 13](https://lacouth.github.io/python_begins-site/unidade6-funcoes/13-decompor-testar/).

Lembre-se do escopo: uma variável criada dentro de uma função não existe
fora dela — quem quiser usar o resultado precisa do `return`. E, se um teste
falhar, o `AssertionError` aponta a linha exata; leia com calma antes de
mexer na função.

---

**Antes de tudo: faça a sua cópia.** *Arquivo → Salvar uma cópia no Drive*.
O caderno aberto direto do repositório é só leitura, e você perde o que
escrever nele.

**Como usar este caderno:** cada exercício tem duas células. Na primeira,
escreva a sua solução no lugar do `# TODO`. A segunda tem os testes —
rode-a e ela diz se a sua função está correta. Não altere a célula de teste.

Se um teste falhar, o Python mostra um `AssertionError` apontando a linha:
é aquele caso específico que a sua função ainda não atende.

**Se algo ficar inexplicável**, use *Ambiente de execução → Reiniciar e
executar tudo*. O caderno guarda o que foi executado na ordem em que você
executou, e não na ordem em que as células aparecem.
